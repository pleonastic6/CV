# ROC und Precision-Recall – Schritt 1

Dieses Notebook ist ein Lernstart, keine vollständige Projektlösung.

**Ziel:** Eine Entscheidungsschwelle auf simulierte Scores anwenden und Fehlalarme von übersehenen Objekten unterscheiden.

Die Daten sind ausdrücklich simuliert. Ein hoher Score bedeutet eine stärkere Einschätzung für die positive Klasse, nicht automatisch eine kalibrierte Wahrscheinlichkeit.

Für diese erste Zelle werden keine zusätzlichen Bibliotheken benötigt. Das Notebook benötigt eine Python-3-Umgebung.


```python
# 1 = gesuchtes Objekt vorhanden, 0 = Objekt fehlt
y_true = [1, 0, 1, 0, 0, 0, 1, 0]

# Simulierte Modell-Scores, keine nachgewiesenen Wahrscheinlichkeiten
scores = [0.95, 0.80, 0.70, 0.60, 0.40, 0.30, 0.20, 0.10]

# Ab dieser Schwelle lautet die Vorhersage: Objekt vorhanden
schwelle = 0.5
y_pred = [int(score >= schwelle) for score in scores]

for nummer, (wahr, score, vorhersage) in enumerate(
    zip(y_true, scores, y_pred), start=1
):
    print(
        f"Bild {nummer}: Wahrheit={wahr}, "
        f"Score={score:.2f}, Vorhersage={vorhersage}"
    )
```
